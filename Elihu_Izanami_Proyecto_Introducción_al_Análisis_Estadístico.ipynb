IMPUTACIÓN

```python
# Verificar valores nulos por columna de texto
columnas_texto = ['director', 'cast', 'country','date_added','rating','duration']
print(df_netflix[columnas_texto].isnull().sum())
```

```python
# Imputación de datos por director
# Es preferible usar una nuevo director que borrar los registros
#Los registros de la columa director tipo text que eran NaN se llenaran con el texto 'sin director'
#Los registros de la columna cast tipo text que eran Nan se llenaran con el texto 'sin cast'
#Los registros de la columna country tipo text que eran Nan se llenaran con el texto 'sin pais'


df_netflix['director'] = df_netflix['director'].fillna('sin director')
df_netflix['cast'] = df_netflix['cast'].fillna('sin cast')
df_netflix['country'] = df_netflix['country'].fillna('sin pais')
df_netflix['rating'] = df_netflix['rating'].fillna('sin rating')
df_netflix['duration'] = df_netflix['rating'].fillna('sin season')
```

```python
#Si las celdas vacías son espacios en blanco: A veces las celdas contienen solo espacios en vez de valores nulos reales (NaN). Reemplázalos por nulos antes de borrar:
#En este caso son solo 10 registros

df_netflix['date_added'] = df_netflix['date_added'].replace(r'^\s*$', np.nan, regex=True)
df_netflix.dropna(subset=['date_added'], inplace=True)
```
